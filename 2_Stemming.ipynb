**STEMMING**

```python
words = ["eating", "going", "gone", "history", "programming", "programs", "finally", "finalize"]
```

**Porter Stemmer**

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
for word in words:
  print(word+"-->"+stemming.stem(word))
```

```python
stemming.stem('congratulations') # disadvantage
```

```python
stemming.stem("sssiting") #disadvantage
```

**RegexpStemmer Class**

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

**Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer
snow_stemmer = SnowballStemmer(language='english')
for word in words:
  print(word+"--->"+snow_stemmer.stem(word))
```

```python
# porter stemmer
print(stemming.stem('fairly'))
print(stemming.stem('sportingly'))

# snowball stemmer --> better than porter stemmer
print(snow_stemmer.stem('fairly'))
print(snow_stemmer.stem('sportingly'))
```
