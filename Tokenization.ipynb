```python
corpus = """Hello, I am nivya.
I am a BTech CSE student. Nivya's Book.
"""
```

```python
corpus
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
## Tokenization
## sentence -> paragraphs
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
for sentence in documents:
  print(sentence)
```

```python
## Tokenization
## Paragraph -> words / Sentence -> words
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
for word in words:
  print(word)
```

```python
## Word punct tokenize --> treats each punctuation as token
## ex: Nivya's --> Nivya, ', s
from nltk.tokenize import wordpunct_tokenize
words = wordpunct_tokenize(corpus)
for word in words:
  print(word)
```

```python
## Tree Bank Word tokenizer
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
words = tokenizer.tokenize(corpus)
for word in words:
  print(word)

## ex: student. (along with full stop is treated as one token), but not for last word
```
